# **ブラックジャック**


## **概要**

本プログラムは、コマンドライン上で遊べるテキストの対戦型トランプゲーム「ブラックジャック」です。プレイヤーとPC（ディーラー）が1対1で対戦し、手札の合計点を「21」を超えない範囲で、できるだけ「21」に近づけることを目指すゲームです。

プログラムが備えている主な機能は以下の通りです。

**カードのランダム配布機能**

トランプの1から13までの数字をランダムに選んで、プレイヤーとディーラーの手札として配ります。

**プレイヤーの操作機能**

キー入力機能（y / n）を取り入れ、「もう1枚カードを引く」か「現在のカードで勝負する」かを自由に選択できます。

**ディーラーの制御機能**

ディーラー側のプログラムとして、手札の合計点が17以上になるまで自動でカードを引き続けるプログラムを入れています。

**勝敗の判定・結果表示機能**

手札が21を超えると失格となる「バースト」や、お互いの合計点を比較して「勝利」「敗北」「引き分け」を判定します。

## **操作方法**

プログラムを実行すると、自動的にプレイヤーとディーラーにカードが2枚ずつ配られます。プレイヤーの手札の合計点と、ディーラーのカード（1枚だけ公開）が画面に表示されます。

**カードを追加するかどうかの選択**

画面に もう1枚引きますか？ (y/n): と表示されたら、キーボードから以下のアルファベットを入力してください。

*y を入力してEnterキーを押す場合（ヒット）*

カードをもう1枚追加で引きます。手札の合計点を確認し、21を超えていなければ何度でも追加で引くことができます。

*n を入力してEnterキーを押す場合（スタンド）*

カードをこれ以上引かず、現在の合計点で勝負します。あなたのターンが終了し、ディーラーのターンへ移ります。

**カードの点数の数え方**

「1 〜 10」のカード = 数字通りの点数として数えます。
「11（J）・12（Q）・13（K）」の絵札 = すべて「10点」としています。

```python
%reset -f
import random as r

def draw_card():
    return r.randint(1, 10)

player_cards = [draw_card(), draw_card()]
dealer_cards = [draw_card(), draw_card()]

print("ブラックジャック開始！")
while True:
  player_sum = sum(player_cards)
  print(f"\nあなたの手札: {player_cards} (合計: {player_sum})")
  print(f"ディーラーのカード: [{dealer_cards[0]}, ?]")

  if player_sum > 21:
    print("21を超えてしまいました（バースト）！")
    break

  choice = input("もう1枚引きますか？ (y/n): ")
  if choice.lower() == 'y':
    player_cards.append(draw_card())
  else:
    break

player_sum = sum(player_cards)

if player_sum <= 21:
    print("\nディーラーのターン")

    while sum(dealer_cards) < 17:
        dealer_cards.append(draw_card())

    dealer_sum = sum(dealer_cards)
    print(f"ディーラーの手札: {dealer_cards} (合計: {dealer_sum})")

    print("\n勝敗は？？？")
    if dealer_sum > 21:
        print("ディーラーがバーストしました！YOU WIN！")
    elif player_sum > dealer_sum:
        print(f"あなた: {player_sum} vs ディーラー: {dealer_sum}   YOU WIN！")
    elif player_sum < dealer_sum:
        print(f"あなた: {player_sum} vs ディーラー: {dealer_sum}   YOU LOSE...")
    else:
        print(f"あなた: {player_sum} vs ディーラー: {dealer_sum}   DRAW")
```
